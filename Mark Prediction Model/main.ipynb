```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from util import showResult

def Machine_learning():
    data = {
        'StudyHR': np.array([1,2,3,4,5,6]),
        'Marks': np.array([100,110,120,130,140,150])
    }

    df = pd.DataFrame(data)

    studyHr = df['StudyHR']
    Marks = df['Marks']

    x = np.array(studyHr).reshape(-1, 1)
    y = np.array(Marks)

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.8, random_state=42)

    showResult(x_train, x_test, y_train, y_test)
    


    model = LinearRegression()
    model.fit(x,y)

    Userinput = float(input("Enter your Hr you've studied: "))
    prediction = model.predict([[Userinput]]) #type: ignore
    print(f"This is the model prediction {prediction[0]:2f}")

Machine_learning()
```
