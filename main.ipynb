```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from util import showResult

# creating def func
def Machine_learning():
    # creating a data frame
    data = {
        'StudyHR': np.array([1,2,3,4,5,6]),
        'Marks': np.array([150,200,250,300,350,400]) 
    }

    df = pd.DataFrame(data)

    studyHr = df['StudyHR']
    Marks = df['Marks']

    
    x = np.array(studyHr).reshape(-1, 1)
    y = np.array(Marks)



    # Training train test split doing for not cheat model
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.8, random_state=42)
    # Show the result as a exam
    showResult(x_train, x_test, y_train, y_test)
    

    # this is the model which we us linear regression
    model = LinearRegression()
    model.fit(x,y)
    # this the userinput and prediction variable to predict the data
    Userinput = float(input("Enter your Hr you've studied: "))
    prediction = model.predict([[Userinput]]) #type: ignore
    print(f"This is the model prediction {prediction[0]:2f}")

Machine_learning()
```
